# 1. Problem and Data Snapshot


**About the Dataset**

Our chosen dataset represents patient portfolios and profiles to simulate real-world healthcare scenarios. Diverse information about everyone is provided, allowing insight into lifestyle, medical background, and demographic factors that might contribute to the overall health of the patient. Examples of these factors include gender, alcohol usage, and the presence of a stroke in medical history. This dataset is formatted as a .csv, containing only tabular, numeric data, with some of the variables being binary. It is a singular file, not yet separated into training and testing data, with 5,000 entries or rows.

**About the Problem**

The goal of diagnostic medicine is to take all necessary information about a patient and come up with a medical diagnosis. "Situated knowledge," a term coined by Donna Haraway, is the idea that all knowledge comes with bias that originates from the knower. We see these biases throughout the history of medicine in the US, contributing to racial disparities in diagnosis, treatment, and patient-provider trust. Our team goal is to train a model that can calculate a health score that would be solely based on medical records, free from bias.

**Target Variable**

Heart_Disease

**Main Input Features**

Age, Gender, BMI, Exercise_Level, Smoking, Alcohol, Blood_Pressure, Cholesterol, Glucose, Fatigue, Chest_Pain, Dizziness, Diabetes, Stroke

**Pairplot**

```python
import matplotlib.patches as mpatches

pairplot_features = ['age', 'bmi', 'blood_pressure', 'cholesterol', 'glucose', 'heart_disease']

pp = sns.pairplot(
    data[pairplot_features],
    hue='heart_disease',
    palette={0: 'steelblue', 1: 'tomato'},
    diag_kind='kde',
    plot_kws={'alpha': 0.5, 'edgecolor': 'none', 's': 15},
    diag_kws={'fill': True, 'alpha': 0.5}
)

no_disease  = mpatches.Patch(color='steelblue', label='No Heart Disease (0)')
yes_disease = mpatches.Patch(color='tomato',    label='Heart Disease (1)')

pp.figure.legend(
    handles=[no_disease, yes_disease],
    title='Heart Disease',
    loc='upper right',
    bbox_to_anchor=(1.02, 1),
    borderaxespad=0,
    frameon=True,
    fontsize=10,
    title_fontsize=11
)
pp.legend.remove()


plt.suptitle('Pairplot of Key Features by Heart Disease Status', y=1.01, fontsize=13)
plt.tight_layout()
plt.show()
```

# 3. Data Quality and Cleaning

##Data Quality Issues
###Missing Values:
**There are no records that contain an n/a, so none are missing.**

###Invalid Zeros:
A zero would be invalid if they exist in the following columns: gender (variable is categorical and either 'Male' or 'Female), bmi, blood pressure, cholesterol, glucose, and health_risk_score.

**There are no unexpected zeros in this dataset.**

###Inconsistent Labels:
We expect the following options for each variable:

age - int 18 - 79

gender - 'Female', 'Male'

bmi - float 0 - 200

exercise_level - int 0, 1, 2

smoking - int 0,1

alcohol - int 0,1

blood_pressure - int 1 - 250

cholesterol - 1 - 500

glucose 1 - 500

fatigue - 0, 1

chest_pain - 0,1

dizziness - 0,1

diabetes - 0,1

stroke - 0,1

heart_disease - 0,1

**Analysis of the data shows that no records do not fall within these labelling expectations.**

###Duplicate Records:
Because there is no unique identifier for each entry, **there is no absolute way to determine if there are duplicate records.**

###Clearly Invalid Entries:
A clearly invalid entry would be a record that is not physically possible given medical knowledge.  Due to the diversity of medical circumstances, we need to be tolerant of many different compositions of the data (unexpected or unusual patient profiles), so unless the value is extreme or labelled incorrectly, we don't truly have an 'invalid' designation.

Therefore, there are **no clearly invalid entries in the dataset.**


###Unusual or Extreme Values:
As mentioned before, we are highly tolerant of unusual values.  However, some expectations can be made with medical knowledge. For example, a glucose level of 700 would be quite inconceivable.  **None such inconceivable, extreme values medically were found.**

# 4. Modeling Readiness Reflection

##Most Important Patterns
The exploratory analysis revealed that lifestyle and health-related features (such as BMI, smoking status, and physical activity) show noticeable relationships with the target variable, Heart_Disease. No single feature appears to dominate prediction, suggesting that a combination of variables will be needed for accurate modeling. Additionally, some features show moderate correlations with each other, indicating potential multicollinearity that may need to be considered in certain models.

##Most Important Data Quality Issues
The dataset is largely clean with no missing values, which is a strong advantage. Outliers are only present when further analysis is done, such as the scatterplots with BMI by heart disease, but not the raw data itself.

##Most Useful Features
Features related to direct health indicators and behaviors appear most useful, including BMI, blood pressure-related variables, and lifestyle factors such as smoking, alcohol use, and physical activity. These variables show clearer patterns in both visualizations and correlation analysis, suggesting they will provide meaningful predictive signal.

##Preprocessing/Modeling Challenges
Key challenges moving forward include handling invalid zero values through imputation or filtering, and applying feature scaling for models sensitive to differences in magnitude. Potential class imbalance in the Heart_Disease variable may require resampling techniques or adjusted evaluation metrics. Additionally, model selection will involve balancing interpretability (such as logistic regression) with predictive performance (such as tree-based models), while also addressing any multicollinearity between features.
